```python
outlier_features = [
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
    "TotalWorkingYears",
    "YearsWithCurrManager"
]

Q1 = df[outlier_features].quantile(0.25)
Q3 = df[outlier_features].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_mask = (
    (df[outlier_features] < lower_bound) |
    (df[outlier_features] > upper_bound)
).any(axis=1)

print("Extreme outliers:", outlier_mask.sum())
# Remove extreme outliers
df_clean = df[~outlier_mask].copy()

print("Original rows:", len(df))
print("Rows after outlier removal:", len(df_clean))
print("Removed rows:", len(df) - len(df_clean))
```

```python
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
```

```python
values = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    values.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(
    range(2, 11),
    values,
    marker="o"
)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_values = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_values.append(score)
```

```python
kmeans_4 = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

labels_4 = kmeans_4.fit_predict(X_scaled)

df_clean["CareerCluster"] = labels_4
df.loc[df_clean.index, "CareerCluster"] = labels_4
cluster_profile_4 = (
    df_clean.groupby("CareerCluster")[career_features]
    .mean()
    .round(2)
)

cluster_profile_4
```

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

hierarchical = AgglomerativeClustering(
    n_clusters=4,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

hierarchical_silhouette = silhouette_score(
    X_scaled,
    hierarchical_labels
)

print(
    "Hierarchical Clustering Silhouette Score:",
    round(hierarchical_silhouette, 4)
)
```

```python
from sklearn.preprocessing import MinMaxScaler

promotion_scaler = MinMaxScaler()

df["PromotionGapNormalized"] = (
    promotion_scaler.fit_transform(
        df[["PromotionGapScore"]]
    ).ravel() * 100
)

df["RetentionOpportunityIndex"] = (
    0.50 * df["PromotionGapNormalized"] +
    0.25 * (df["TrainingNeedIndicator"] * 100) +
    0.25 * (df["ManagerStabilityImpact"] * 100)
)

eligible = (
    (df["Attrition"] == 0) &
    (df["JobSatisfaction"] >= 3)
)

df.loc[~eligible, "RetentionOpportunityIndex"] = 0

df["RetentionOpportunityIndex"] = (
    df["RetentionOpportunityIndex"].round(2)
)
```
